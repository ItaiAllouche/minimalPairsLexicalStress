```python
import numpy as np
import os
import matplotlib.pyplot as plt

from utils.lrp import do_lrp
from utils.loader import *

def run():
    checkpoint_path = f'./checkpoints/checkpoint_VGG16/ckpt.t7'
    model, device = load_model(checkpoint_path=checkpoint_path)

    # iterate over examples files and generate spectogram and corresponded LRP heatmap
    for dir in os.listdir('./examples'):
        dir_path = os.path.join('./examples', dir)
        for wav in os.listdir(dir_path):
            wav_path = os.path.join(dir_path, wav)
            
            spect, spect_dB = spect_loader(wav_path)
            heatmap = do_lrp(model=model, spect=spect, device=device)
            spect_dB = spect_dB.unsqueeze(1)
            spect_dB = spect_dB.squeeze().cpu().detach().numpy()
            spect_dB = np.expand_dims(spect_dB, axis=-1)
            
            fig, axes = plt.subplots(1, 2, figsize=(9, 3), constrained_layout=True)
            axes[0].imshow(spect_dB[:117, :50], extent=[0,500,0,8000], origin='lower' ,aspect='auto')
            axes[0].set_title('Spectogram', fontsize=12)
            axes[0].set_xlabel('Time [ms]', fontsize=10)
            axes[0].set_ylabel('Frequncey [Hz]', fontsize=10)

            axes[1].imshow(spect_dB[:117, :50], extent=[0,500,0,8000], cmap='gray', origin='lower', aspect='auto',vmin=-1.2)
            axes[1].imshow(heatmap, alpha=0.8, extent=[0,500,0,8000], cmap='Greens', vmin=0, vmax=0.2, origin='lower' ,aspect='auto')
            axes[1].set_title('LRP Heatmp', fontsize=12)
            axes[1].set_xlabel('Time [ms]', fontsize=10)
            axes[1].set_ylabel('Frequncey [Hz]', fontsize=10)    

            plt.show()

if __name__ == '__main__':
    run()
```
